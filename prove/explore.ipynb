```python
import sys
import os
import pandas as pd

# Aggiungiamo il percorso del progetto al path di Python
project_path = '/Users/sal/Desktop/Programming Exam Project'
if project_path not in sys.path:
    sys.path.append(project_path)

from data_loader import load_team_stats

# 1. Imposta la stagione
selected_season = 2026

# 2. Scarichiamo le 4 tabelle dal Data Loader aggiornato
df_team_trad, df_team_adv, df_opp_trad, df_opp_adv = load_team_stats(selected_season)


cols_to_drop = ['team.name', 'team.imageUrl', 'team.tvCodes', 'teamRanking', 'gamesPlayed']
df_adv_clean = df_team_adv.drop(columns=[c for c in cols_to_drop if c in df_team_adv.columns])
# 3. Uniamo le statistiche della squadra (Tradizionali + Avanzate)
df_team_full = pd.merge(
    df_team_trad, 
    df_adv_clean, 
    on='team.code'
)

# 4. Puliamo i dataframe degli avversari rimuovendo le colonne descrittive duplicate
df_opp_trad_clean = df_opp_trad.drop(columns=[c for c in cols_to_drop if c in df_opp_trad.columns])
df_opp_adv_clean = df_opp_adv.drop(columns=[c for c in cols_to_drop if c in df_opp_adv.columns])

# 5. Eseguiamo i merge con le statistiche degli avversari pulite
df_team_full = pd.merge(
    df_team_full, 
    df_opp_trad_clean, 
    on='team.code', 
    suffixes=('', '_opp_trad')
)

df_team_full = pd.merge(
    df_team_full, 
    df_opp_adv_clean, 
    on='team.code', 
    suffixes=('', '_opp_adv')
)

# 6. Stampiamo l'elenco ordinato di tutte le colonne risultanti
print(f"=== TOTALE COLONNE PRESENTI: {len(df_team_full.columns)} ===\n")
for idx, col in enumerate(df_team_full.columns, 1):
    print(f"{idx}. {col}")
```

```python
# Controllo rapido in Jupyter per vedere se esistono stringhe di colonna esattamente identiche
duplicates = df_team_full.columns[df_team_full.columns.duplicated()].tolist()
print("Colonne duplicate esattamente:", duplicates)
```
